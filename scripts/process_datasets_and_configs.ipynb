## Add parameters to prices.csv
Some of the parameters in `prices.csv` make the inputs for calculating species prices. E.g.:

$$
S_p(S_c, c) = \operatorname{price}
$$

where $S$ is the species, $S_p$ is the species price, $S_c$ is the species code and $c$ is the category_code. However, these parameters can change throughout the project's lifetime, so we must be able to rapidly adapt to the new requirements. The following tool helps adding new parameters to the `UpdateSales` messages. It will go through each file in the corresponding directory and add the `category_code` field. To use correct formatting, be sure to go to Bruno and save each message.

Strategy:
1. Take the first UpdateSales request and add the category codes.
2. For each of the other requests, use the same request as the first one, but update the timestamp.

### First file

```python
update_dir = Path("../grpc_requests/UpdateSalesByTicks")

first_file_data = None

# Update the first file
print(f"Processing file: UpdateSales1_(2013_01).yml")
with open(Path(update_dir, "UpdateSales1_(2013_01).yml"), "r") as f:
    first_file_data = yaml.safe_load(f)

msg = first_file_data.get("grpc").get("message")
json_data = json.loads(msg)
market_sales = json_data.get("sales_summary").get("market_sales")

categories = [
    "Fresh - Whole"
]

for ms in market_sales:
    for sale in ms.get("sales"):
        rand_index = random.randint(0, len(categories)-1)
        sale["category_code"] = categories[rand_index]

json_data["sales_summary"] = {
    "market_sales": market_sales
}

first_file_data.get("grpc")["message"] = json.dumps(json_data, indent=4)

with open(Path(update_dir, 'UpdateSales1_(2013_01).yml'), 'w') as outfile:
    #  default_flow_style=False, 
    yaml.dump(first_file_data, outfile, sort_keys=False)
```

### The rest of the files

```python

# Update the other files' timestamps
for i, file in enumerate(update_dir.iterdir()):
    if file.stem == "folder" or file.stem == "UpdateSales1_(2013_01)":
        continue
    print(f"Processing file: {file}")

    with open(file, "w") as f:
        yaml.dump(first_file_data, f, sort_keys=False)

print("Complete")
```
